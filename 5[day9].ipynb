```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class linkedlist:
    def __init__(self):
        self.head = None

    def insert(self, data):
        newnode = Node(data)
        if self.head is None:   # fixed syntax
            self.head = newnode
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = newnode

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end="->")
            temp = temp.next   # fixed indentation


l = linkedlist()
l.insert(100)
l.insert(200)
l.insert(300)
l.insert(400)
l.display()
```
