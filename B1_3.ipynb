```python
k = int(input())
arr = list(map(int, input().split()))
max_length = 0
for i in range(len(arr)):
    current_sum = 0
    for j in range(i, len(arr)):
        current_sum += arr[j]
        if current_sum <= k:
            max_length = max(max_length, j - i + 1)
print(max_length)
```

```python
k = int(input())
arr = list(map(int, input().split()))
left = 0
current_sum = 0
max_length = 0
for right in range(len(arr)):
    current_sum += arr[right]
    while current_sum > k:
        current_sum -= arr[left]
        left += 1
    max_length = max(max_length, right - left + 1)
print(max_length)
```

sliding window

```python
s=input()
left=0
max_length=0
seen=set()
for right in range(len(s)):
  while s[right] in seen:
      seen.remove(s[left])
      left+=1
  seen.add(s[right])
  max_length=max(max_length,right-left+1)
print(max_length)
```

```python
s = input()
left = 0
max_length = 0
last_seen = {}
for right in range(len(s)):
    if s[right] in last_seen and last_seen[s[right]] >= left:
        left = last_seen[s[right]] + 1
    last_seen[s[right]] = right
    max_length = max(max_length, right - left + 1)
print(max_length)
```

find the length of the longest subarray with consecutive ones

```python
n = int(input())
arr = list(map(int, input().split()))
count = 0
max_count = 0
for i in range(n):
    if arr[i] == 1:
        count += 1
        max_count = max(max_count, count)
    else:
        count = 0
print(max_count)
```
